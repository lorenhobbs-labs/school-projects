## Import Key Libraries

```python
## Please write and run your code here
# Please add cells as needed
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Import the Data

```python
## Please write and run your code here
# Please add cells as needed
df = pd.read_csv('./die_rolls.csv')
```

## Visualize the data frame

```python
## Please write and run your code here
# Please add cells as needed
# Inspect the first few rows and structure of the dataset
print("DataFrame Shape:", df.shape)
print("\nFirst 5 rows:")
print(df.head())
```

## Interpret the results
#### Question 1: 
How many sides did the dice have, $K$?

```python
# Compute the number of sides (K) from the columns
K = df.shape[1]
print(f"Question 1: The dice has K = {K} sides.")
```

#### Question 2:
How many rolls, $N$, were there?

```python
# Compute the total number of rolls (N) from the rows
N = df.shape[0]
print(f"Question 2: There were N = {N} total rolls.")
```

## Compute m and the MLE

```python
## Please write and run your code here
# Please add cells as needed
# 1. Calculate the observed counts (m_k) for each face
# Summing down each column gives the total 1s per face
m = df.sum(axis=0).values

# Display counts
for idx, count in enumerate(m, start=1):
    print(f"m_{idx}: {count} times")

# 2. Calculate the Maximum Likelihood Estimation (MLE) for all faces
mle = m / N

print("\n--- MLE Calculations ---")
for idx, val in enumerate(mle, start=1):
    print(f"MLE for mu_{idx}: {val:.4f}")

# 3. Specific Answer Outputs for remaining questions
most_often_idx = np.argmax(m) + 1
smallest_mle = np.min(mle)
analytic_val = 1 / K
abs_diff = abs(smallest_mle - analytic_val)
sum_mle = np.sum(mle)

print("\n--- Assignment Answer Summary ---")
print(f"Question 3 (MLE for mu_2): {mle[1]:.4f}")
print(f"Question 4 (Most frequent face k): {most_often_idx}")
print(f"Question 5 (Times rolled): {m[most_often_idx-1]}")
print(f"Question 6 (Smallest MLE): {smallest_mle:.4f}")
print(f"Question 7 (Absolute deviation from analytic): {abs_diff:.4f}")
print(f"Question 8 (Sum of all MLEs): {sum_mle:.4f}")
```

#### Display m

```python
## Please write and run your code here
# Please add cells as needed
# Sum each column to get the total count for each face
m = df.sum(axis=0).values
print("Counts vector m:", m)
```

#### Display the MLE of mu

```python
## Please write and run your code here
# Please add cells as needed
# Calculate MLE by dividing counts by total rows N
mle = m / df.shape[0]
print("MLE vector mu:")
for idx, val in enumerate(mle, start=1):
    print(f"  mu_{idx}: {val:.4f}")
```

#### Question 3:
What was the MLE for $\mu_2$? 

Please express your answer to four decimal places.

```python
# Extract the MLE for the second face (index 1 in 0-based indexing)
print(f"{mle[1]:.4f}")
```

#### Question 4:
What number, $k$, was rolled the most often?

```python
# Find the face (1-indexed) with the maximum count
most_often_k = np.argmax(m) + 1
print(most_often_k)
```

#### Question 5:
How many times was that number rolled?

```python
# Extract the maximum count from vector m
max_rolls = np.max(m)
print(max_rolls)
```

#### Question 6:
What was the smallest MLE?

Please express your answer to four decimal places.

```python
# Find the minimum MLE value across all faces
smallest_mle = np.min(mle)
print(f"{smallest_mle:.4f}")
```

#### Question 7: 
How far was that value from the analytic value of $\mu_{MLE}$? 

Please take the absolute value and round to 4 decimal places.

```python
# Calculate the absolute difference from the fair analytic value (1/K)
K = df.shape[1]
analytic_val = 1 / K
abs_diff = abs(smallest_mle - analytic_val)
print(f"{abs_diff:.4f}")
```

#### Question 8:
What is $\displaystyle\sum\limits_{k = 1}^{K} \mu_{(mle,k)}$?

```python
# Sum all the elements in the MLE vector
sum_mle = np.sum(mle)
print(f"{sum_mle:.4f}")
```

#### Question 9:
Do you think that this is a fair die? Please explain your reasoning.

```python
# Print out metrics side by side to evaluate fairness
K = df.shape[1]
analytic_val = 1 / K

print(f"Theoretical Fair Value (1/K): {analytic_val:.4f}\n")
print("Face | Observed MLE | Absolute Deviation")
print("-" * 40)
for idx, val in enumerate(mle, start=1):
    dev = abs(val - analytic_val)
    print(f" {idx:2d}  |    {val:.4f}    |     {dev:.4f}")

# Markdown explanation guide:
# Check if the deviations are small random fluctuations (e.g., < 0.02)
# or if any single face deviates massively from the fair value.
```

```python
Yes, this is a fair die.Reasoning:Uniform Baseline: For a perfectly balanced 20-sided die (\(K = 20\)), the theoretical probability of rolling any individual face is exactly \(\frac{1}{20} = 0.0500\).Minimal Deviations: The empirical Maximum Likelihood Estimators (\(\^{\mu }_{k}\)) calculated from the data hover tightly around this baseline across all 20 faces.Statistical Bounds: The largest observed absolute deviation in the entire dataset is a minuscule \(0.0062\) (on Face 12).Conclusion: These tight results indicate that the minor differences between the observed frequencies and the theoretical mean are caused strictly by standard random sampling noise. There is no evidence of a systematic bias or a loaded die.
```
